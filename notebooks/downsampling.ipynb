## Figure downsampling

```python
#| label: figDownsamplingX

import numpy as np
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.io as pio
from scipy.io import loadmat

# Interactive display in the notebook
pio.renderers.default = "plotly_mimetype"

# Check if static image export is available for PDF
try:
    pio.to_image(go.Figure(), format="png")
    static_export = True
except Exception as exc:
    static_export = False
    print(f"Static PNG export unavailable: {type(exc).__name__}: {exc}")

#K-spaces and image spaces downsampled by half according to x
data = loadmat("kx.mat")
kx = data["magx"]

data = loadmat("magimx.mat")
magimx = data["magimx"]

fig = make_subplots(1, 2, subplot_titles=("Downsampled k-space accoridng to X", "Reconstructed image"))

fig.add_trace(go.Heatmap(visible=True, z=kx[:,:]), row=1, col=1)
fig.add_trace(go.Heatmap(visible=True, z=magimx[:,:]), row=1, col=2)

fig
```
