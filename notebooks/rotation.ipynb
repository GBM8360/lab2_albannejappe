## Figure head rotation

```python
#| label: figRotation

import numpy as np
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.io as pio
from scipy.io import loadmat

# Interactive display in the notebook
pio.renderers.default = "plotly_mimetype"

# Check if static image export is available for PDF
try:
    pio.to_image(go.Figure(), format="png")
    static_export = True
except Exception as exc:
    static_export = False
    print(f"Static PNG export unavailable: {type(exc).__name__}: {exc}")

data = loadmat("kStack.mat")
kStack = data["kStack"]

data = loadmat("phaseStack.mat")
phaseStack = data["phaseStack"]

data = loadmat("magStack.mat")
magStack = data["magStack"]

n = kStack.shape[2]

fig = make_subplots(1, 3, subplot_titles=("Assembled k-space", "Magnitude of the reconstructed image", "Phase of the reconstructed image"))

for i in range(n):
    fig.add_trace(go.Heatmap(visible=False, z=kStack[:, :, i]), row=1, col=1)
    fig.add_trace(go.Heatmap(visible=False, z=magStack[:, :, i]), row=1, col=2)
    fig.add_trace(go.Heatmap(visible=False, z=phaseStack[:, :, i]), row=1, col=3)

DEFAULT = 0
fig.data[0].visible = True
fig.data[1].visible = True
fig.data[2].visible = True

steps = []
for i in range(n):
    visible = [False] * (3*n)
    visible[3*i] = True       # k-space
    visible[3*i + 1] = True   # mag
    visible[3*i + 2] = True   # phase
    steps.append(dict(method="update", args=[{"visible": visible}], label=str(i)))
fig.update_layout(
    sliders=[dict(active=DEFAULT, steps=steps, currentvalue={"prefix": "Head angle of rotation: "})]
)

fig


```
