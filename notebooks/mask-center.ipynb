## Figure mask center

```python
#| label: figMaskCenter

import numpy as np
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.io as pio
from scipy.io import loadmat

try:
    pio.to_image(go.Figure(), format="png")
    pio.renderers.default = "plotly_mimetype+png"
except Exception as exc:
    pio.renderers.default = "plotly_mimetype"
    print(f"Static PNG export unavailable ({type(exc).__name__}); "
          "the website is unaffected, but PDF figures will be captions only.")

data = loadmat("maskStack.mat")
maskStack = data["maskStack"]

data = loadmat("imgStack.mat")
imgStack = data["imgStack"]

n = maskStack.shape[2]

fig = make_subplots(1, 2, subplot_titles=("Masked k-space", "Reconstructed image"))

for i in range(n):
    fig.add_trace(go.Heatmap(visible=False, z=maskStack[:, :, i]), row=1, col=1)
    fig.add_trace(go.Heatmap(visible=False, z=imgStack[:, :, i]), row=1, col=2)

DEFAULT = 0
fig.data[0].visible = True
fig.data[1].visible = True

'''frames = []
for i in range(maskStack.size(2)):
    frames.append(go.Frame(
        data=[go.Heatmap(z=maskStack[:, :, i]), go.Heatmap(z=imageStack[:, :, i])],
        name=str(i)
        )
    )
fig.frames = frames'''

steps = []
for i in range(n):
    visible = [False] * (2*n)
    visible[2*i] = True       # k-space
    visible[2*i + 1] = True   # image
    steps.append(dict(method="update", args=[{"visible": visible}], label=str(i)))
fig.update_layout(
    sliders=[dict(active=DEFAULT, steps=steps, currentvalue={"prefix": "Mask size: "})]
)

fig

```
